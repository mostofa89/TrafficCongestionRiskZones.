```python
df = df.drop(
    columns = [
        "ID",
        "Source",
        "End_Lat",
        "End_Lng",
        "End_Time",
        "Distance(mi)",
        "Description",
        "Street",
        "Zipcode",
        "Airport_Code",
        "Country",
        "County",
        "City",
        "Turning_Loop",
        "Weather_Timestamp",
        "Timezone"
        
    ]
)
df.head(10)
```

```python
df["Start_Time"] = pd.to_datetime(df["Start_Time"], format="mixed", errors="coerce")
df["Hour"] = df["Start_Time"].dt.hour
df["Month"] = df["Start_Time"].dt.month
df["DayOfWeek"] = df["Start_Time"].dt.dayofweek
df = df.drop(columns="Start_Time")

# numbers -> median, text -> "Unknown"
for c in df.select_dtypes("number").columns:
    df[c] = df[c].fillna(df[c].median())
for c in df.select_dtypes(exclude="number").columns:
    df[c] = df[c].fillna("Unknown")
```

```python
df.head(10)
```

```python
bool_cols = df.select_dtypes("bool").columns

bool_cols
```

```python
df[bool_cols] = df[bool_cols].astype(int)

df.head(10)
```

```python
object_cols = df.select_dtypes("object").columns

object_cols
```

```python
twi_cols = ["Sunrise_Sunset","Civil_Twilight","Nautical_Twilight","Astronomical_Twilight"]
for c in twi_cols:
    df[c] = df[c].map({"Day": 1, "Night": 0}).fillna(0).astype(int)
```

```python
df["Wind_Direction"] = df["Wind_Direction"].replace(
    {"CALM":"Calm","VAR":"Variable","North":"N","South":"S","East":"E","West":"W"}
).fillna("Unknown")

df["Weather_Condition"] = df["Weather_Condition"].fillna("Unknown")
```

```python
from sklearn.preprocessing import LabelEncoder

for c in ["State", "Wind_Direction", "Weather_Condition"]:
    df[c] = LabelEncoder().fit_transform(df[c].astype(str))

print(df[["State","Wind_Direction","Weather_Condition"]].head())
```

```python
df = pd.get_dummies(df, columns=["State","Wind_Direction","Weather_Condition"],
                    drop_first=True, dtype=int)
```

```python
from sklearn.model_selection import train_test_split

df["Severity"] = df["Severity"] - 1          # classes 0-3 (skip if already done)
X = df.drop(columns="Severity")
y = df["Severity"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42)
print(X_train.shape, X_test.shape)
```
